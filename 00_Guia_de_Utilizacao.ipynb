## 3. Credenciais Necessárias

Os notebooks de download e streaming precisam de variáveis de ambiente para S3/MinIO.

Backend LAZ:

```text
AWS_ENDPOINT_URL1
AWS_ACCESS_KEY_ID1
AWS_SECRET_ACCESS_KEY1
```

Backend MDT/MDS:

```text
AWS_ENDPOINT_URL2
AWS_ACCESS_KEY_ID2
AWS_SECRET_ACCESS_KEY2
```

Estes são os nomes usados nos exemplos dentro da VM.


````python
import os

checks = {
    "Bucket": ["AWS_S3_BUCKET"],
    "Endpoint LAZ": ["AWS_ENDPOINT_URL1"],
    "Access key LAZ": ["AWS_ACCESS_KEY_ID1"],
    "Secret key LAZ": ["AWS_SECRET_ACCESS_KEY1"],
    "Endpoint MDT/MDS": ["AWS_ENDPOINT_URL2"],
    "Access key MDT/MDS": ["AWS_ACCESS_KEY_ID2"],
    "Secret key MDT/MDS": ["AWS_SECRET_ACCESS_KEY2"],
}

for label, names in checks.items():
    found = [name for name in names if os.getenv(name)]
    if found:
        print(f"OK        {label}: {', '.join(found)}")
    else:
        print(f"em falta  {label}: nenhum de {', '.join(names)}")
````
